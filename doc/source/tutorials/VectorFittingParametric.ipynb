# Parametric Vector Fitting
The implementation for the parametric vector fitting is based on the algorithm proposed in [1]

## Mathematical Description
The parametric model function $\mathbf{\underline{H}}(\mathrm{\underline{s}; \lambda})$ is defined in the Laplace domain with $\mathrm{\underline{s}} = \sigma + \mathrm{j} \omega$ and for a parameter (or a vector of parameters) $\lambda$:
\begin{equation}
\mathbf{\underline{H}}(\mathrm{\underline{s}; \lambda}) = \frac{\mathbf{Q}_0(\lambda) + \sum_{k=1}^K \frac{\mathbf{Q}_k(\lambda)}{\mathrm{\underline{s}}-\underline{p}_k}}{\mathbf{R}_0(\lambda) + \sum_{k=1}^K \frac{\mathbf{R}_k(\lambda)}{\mathrm{\underline{s}}-\underline{p}_k}}
\end{equation}
This model is based on a set of $K$ poles $\underline{p}_k$, which do not depend on the value(s) of the parameter(s). In contrast, the numerator coefficients $\mathbf{Q}_k(\lambda)$ and the denominator coefficients $\mathbf{R}_k(\lambda)$ do depend on the parameters. They get calculated on a given grid of parameter samples using regular vector fitting. These coefficients can then be interpolated to calculate the model response at any parameter value between the gridpoints.
