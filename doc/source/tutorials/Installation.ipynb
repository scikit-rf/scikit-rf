# Installation

scikit-rf requires Python 3.10 or newer.

## pip

Using the [Python Package Index (PyPI)](https://pypi.org/):

```shell
pip install scikit-rf
```

## uv

Using [uv](https://docs.astral.sh/uv/):

```shell
uv add scikit-rf
# or in an activated virtual environment:
uv pip install scikit-rf
```

## anaconda

Using [conda-forge](https://conda-forge.org/):

```shell
conda install -c conda-forge scikit-rf
```

## Optional extras

Some features need extra packages, which you can install with an extra:

| Extra | Installs | Needed for |
|---|---|---|
| `plot` | matplotlib | plotting (`Network.plot_*`, `skrf.plotting`) |
| `visa` | PyVISA, pyvisa-py | instrument control (`skrf.vi`) |
| `xlsx` | openpyxl | Excel export (`write_spreadsheet`) |
| `netw` | networkx | `Circuit`'s graph functions |

Using pip:

```shell
pip install "scikit-rf[plot]"
pip install "scikit-rf[plot,visa]"   # several at once, e.g. for the GUI apps
```

Using uv:

```shell
uv add "scikit-rf[plot]"
```

## From source

To get the development version from [GitHub](https://github.com/scikit-rf/scikit-rf):

```shell
git clone https://github.com/scikit-rf/scikit-rf.git
cd scikit-rf
pip install -e .
# or with uv:
uv sync
```

## Instrument drivers

PyVISA-py, installed by the `visa` extra, is enough to talk to many instruments. Some setups need the vendor's VISA library instead: for example, Keysight's N5227B PNA over GPIB needed the [Keysight IO Libraries Suite](https://www.keysight.com/us/en/lib/software-detail/computer-software/io-libraries-suite-downloads-2175637.html). In that case, point the virtual instrument at that backend; see the [PyVISA backends documentation](https://pyvisa.readthedocs.io/en/latest/advanced/backends.html). If you still run into trouble, please open an issue with the details.
